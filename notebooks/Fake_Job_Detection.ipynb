## NLP (Natural Language Processing)


```python
import re                                   #Regular Expressions (Use - text cleaning)
from nltk.tokenize import word_tokenize     #Sentence → Words
from nltk.corpus import stopwords           #common words that usually don't help prediction. (and,is,the..)

stop_words = set(stopwords.words("english"))       #Creates a collection of stopwords.

def preprocess_text(text):
    text = str(text).lower()

    text = re.sub(r"[^a-zA-Z]", " ", text)    #removes numbers, symbols, punctuation, etc.

    tokens = word_tokenize(text)

    tokens = [word for word in tokens if word not in stop_words]

    return " ".join(tokens)    #Convert list back into text.
```

```python
clean_tokens = []                #Creates an empty list

for word in tokens:              #Go through every word
    if word not in stop_words:   #Is this word a stopword? -If No, Keep it. -/If Yes, Ignore it.

        clean_tokens.append(word) #Store useful words.

print(clean_tokens)
```

Lemmatization

```python
lemmatized_words = []

for word in clean_tokens:
    lemmatized_words.append(lemmatizer.lemmatize(word))

print(lemmatized_words)           

#running, runs, ran = after lemmatized - ran   - //noun by default
```

Join Words Again

```python
clean_text = " ".join(lemmatized_words)

print(clean_text)
```

combined function

```python
# Build the Function

def preprocess_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = re.sub(r'[^\w\s]', '', text)

    # Tokenize
    tokens = word_tokenize(text)

    # Remove stopwords
    tokens = [word for word in tokens if word not in stop_words]

    # Lemmatize
    tokens = [lemmatizer.lemmatize(word) for word in tokens]

    # Join words back into a sentence
    return " ".join(tokens)
```

## Comparison of all models 

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# 9 Models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier
from sklearn.svm import LinearSVC
from sklearn.naive_bayes import MultinomialNB
from sklearn.neighbors import KNeighborsClassifier
from xgboost import XGBClassifier

# -------------------------------------------------------------
# 1. Load Dataset & Prepare Data
# -------------------------------------------------------------
# Load the cleaned dataset
if os.path.exists('data/processed/cleaned_fake_jobs.csv'):
    df = pd.read_csv('data/processed/cleaned_fake_jobs.csv')
else:
    df = pd.read_csv('../data/processed/cleaned_fake_jobs.csv')

# Determine the text and target column names
text_col = 'clean_text' if 'clean_text' in df.columns else ('text' if 'text' in df.columns else 'description')
target_col = 'fraudulent'

X = df[text_col].fillna('')
y = df[target_col]

# Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# TF-IDF Vectorization
tfidf = TfidfVectorizer(max_features=5000, ngram_range=(1, 2), stop_words='english')
X_train_vec = tfidf.fit_transform(X_train)
X_test_vec = tfidf.transform(X_test)

# -------------------------------------------------------------
# 2. Define the 9 Models
# -------------------------------------------------------------
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "SVM (Linear)": LinearSVC(random_state=42),
    "Multinomial Naive Bayes": MultinomialNB(),
    "KNN": KNeighborsClassifier(),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
    "AdaBoost": AdaBoostClassifier(random_state=42),
    "XGBoost": XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)
}

# -------------------------------------------------------------
# 3. Train & Evaluate in a Single Loop
# -------------------------------------------------------------
results = []
for name, model in models.items():
    model.fit(X_train_vec, y_train)
    y_pred = model.predict(X_test_vec)
    
    results.append({
        "Model": name,
        "Accuracy": round(accuracy_score(y_test, y_pred), 4),
        "Precision": round(precision_score(y_test, y_pred, zero_division=0), 4),
        "Recall": round(recall_score(y_test, y_pred, zero_division=0), 4),
        "F1-Score": round(f1_score(y_test, y_pred, zero_division=0), 4)
    })

# -------------------------------------------------------------
# 4. Display Comparison Table
# -------------------------------------------------------------
comparison_df = pd.DataFrame(results).sort_values(by="F1-Score", ascending=False).reset_index(drop=True)
print("\n--- 9 Models Performance Comparison ---")
print(comparison_df.to_string(index=False))

# -------------------------------------------------------------
# 5. Plot Performance Comparison Bar Chart
# -------------------------------------------------------------
plt.figure(figsize=(14, 6))
df_melted = comparison_df.melt(id_vars="Model", var_name="Metric", value_name="Score")

sns.barplot(data=df_melted, x="Model", y="Score", hue="Metric", palette="Set2")
plt.title("Performance Comparison of All Algorithms", fontsize=14, fontweight='bold')
plt.ylabel("Score (0 to 1)", fontsize=12)
plt.ylim(0.0, 1.1)
plt.xticks(rotation=30, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import joblib

# 1. Get the trained XGBoost model from your dictionary
best_model = models["XGBoost"]

# 2. Save model and vectorizer artifacts
joblib.dump(best_model, 'best_fake_job_model_xgb.pkl')
joblib.dump(tfidf, 'tfidf_vectorizer.pkl')

print("✅ Best model (XGBoost) and TF-IDF vectorizer saved successfully!")
```
